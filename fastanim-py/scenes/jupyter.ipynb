# fastanim in Jupyter

A cell that ends in a `@scene` function, or a timeline from `.bake()`, plays the scene under it. Run `just jupyter` from the repository root to open this notebook.

```python
from fastanim import *
```

## A scene written here

```python
@scene
def hello(s):
    title = s.add(text("Hello, fastanim").to_edge(UP))
    s.play(write(title))
    ring = s.add(circle(1.5, stroke=BLUE, stroke_width=0.06))
    s.play(create(ring), run_time=1.5)
    eq = s.add(math_tex("a^2 + b^2 = c^2").scale(1.5).to_edge(DOWN))
    s.play(write(eq))
    s.play(transform_diff(eq, math_tex("a^2 = c^2 - b^2").scale(1.5).to_edge(DOWN)))

hello
```

Bake once to read the timeline and show it; `display` shows it from anywhere in a cell.
